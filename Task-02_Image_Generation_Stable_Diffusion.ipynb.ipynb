# Task-02: Image Generation with Pre-trained Models using Stable Diffusion

## Objective

To generate images from text prompts using a pre-trained Stable Diffusion model.
The project demonstrates how text prompts can be converted into AI-generated images
using a pre-trained generative AI model.

## 1. Libraries and Environment

The project was implemented using Python in Google Colab with an NVIDIA T4 GPU.

### Libraries Used

- PyTorch – for deep learning and GPU computation
- Diffusers – for using the pre-trained Stable Diffusion model
- Transformers – for transformer-based text processing
- Accelerate – for efficient model execution on GPU

### Hardware

- GPU: NVIDIA T4
- Platform: Google Colab

```python
!pip install -q diffusers transformers accelerate
```

## 2. Loading the Pre-trained Stable Diffusion Model

Stable Diffusion is a pre-trained generative AI model that can create images
from natural language text prompts.

In this project, the pre-trained Stable Diffusion v1.5 model was used.
The model was loaded using the Hugging Face Diffusers library and executed
on the NVIDIA T4 GPU available in Google Colab.

```python
import torch
from diffusers import StableDiffusionPipeline

model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")

print("Stable Diffusion model loaded successfully! ✅")
```

## 3. Experiments

### Experiment 1: Cute Robot Learning Python

**Prompt:**  
"A cute robot learning Python in a classroom"

This experiment demonstrates basic text-to-image generation using Stable Diffusion.

```python
prompt = "A cute robot learning Python in a classroom"

image = pipe(prompt).images[0]

display(image)

image.save("generated_images/experiment_1_robot.png")

print("Experiment 1 image saved successfully! ✅")
```

### Experiment 2: Futuristic City

**Prompt:**  
"A futuristic city with flying cars, modern skyscrapers, neon lights, cinematic digital art"

This experiment demonstrates the generation of a futuristic environment
using a descriptive text prompt.

```python
prompt = """
A futuristic city with flying cars,
modern skyscrapers, neon lights,
cinematic digital art
"""

image = pipe(prompt).images[0]

display(image)

image.save("generated_images/experiment_2_futuristic_city.png")

print("Experiment 2 image saved successfully! ✅")
```

### Experiment 3: Peaceful Mountain Village

**Prompt:**  
"A peaceful mountain village surrounded by green forests, a beautiful river flowing through the valley, sunrise, realistic landscape photography"

This experiment demonstrates realistic landscape image generation.

```python
prompt = """
A peaceful mountain village surrounded by green forests,
a beautiful river flowing through the valley,
sunrise, realistic landscape photography
"""

image = pipe(prompt).images[0]

display(image)

image.save("generated_images/experiment_3_mountain_village.png")

print("Experiment 3 image saved successfully! ✅")
```

### Experiment 4: Watercolor Robot

**Prompt:**  
"A small robot standing in a modern city, watercolor painting style, soft colors, artistic illustration"

This experiment demonstrates how a text prompt can be used to control
the artistic style of the generated image.

```python
prompt = """
A small robot standing in a modern city,
watercolor painting style, soft colors,
artistic illustration
"""

image = pipe(prompt).images[0]

display(image)

image.save("generated_images/experiment_4_watercolor_robot.png")

print("Experiment 4 image saved successfully! ✅")
```

### Experiment 5: AI Healthcare Laboratory

**Prompt:**  
"An advanced artificial intelligence healthcare laboratory, a doctor and an AI robot analyzing medical data, modern medical equipment, futuristic environment, realistic digital art"

This experiment demonstrates image generation for a healthcare and
artificial intelligence themed scenario.

```python
prompt = """
An advanced artificial intelligence healthcare laboratory,
a doctor and an AI robot analyzing medical data,
modern medical equipment, futuristic environment,
realistic digital art
"""

image = pipe(prompt).images[0]

display(image)

image.save("generated_images/experiment_5_ai_healthcare.png")

print("Experiment 5 image saved successfully! ✅")
```
